# Europe 2026 catalogue: brief-paper figures

Builds the figures for a short paper describing
`data/heatwave_events_europe_2026.csv`, the 32-event CTX90pct catalogue from
`scripts/detect_heatwaves_europe.py`. Each writes a PDF to `plots/detection/`.

| Figure | File | Shows |
|---|---|---|
| 1 | `europe_2026_fig1_map.pdf` | Where the events were, over heat-wave days per cell |
| 1b | `europe_2026_fig1b_map_borders.pdf` | The same map over national borders |
| 2 | `europe_2026_fig2_timeline.pdf` | When they ran across the season |
| 3 | `europe_2026_fig3_severity.pdf` | How severity is distributed across the 32 |
| 4 | `europe_2026_fig4_anomaly_vs_tmax.pdf` | Absolute heat against heat relative to local climate |
| 5 | `europe_2026_fig5_koppen.pdf` | Which climates the catalogue is made of |
| 6 | `europe_2026_fig6_footprints.pdf` | The footprint of each of the 8 most severe events |
| 7 | `europe_sensitivity_ranks.pdf` | What survives a change of detector settings, 2003 / 2022 / 2026 |
| 7b | `europe_sensitivity_rank_changes.pdf` | The same result as a rank-change grid; stable seasons read as blank |

Runs on CPU in the repo `.venv`, about 15 seconds end to end. Needs
`data/era5_europe/europe_tmax_2026.nc` and the cached threshold, both of which
are gitignored, so regenerate them with `scripts/download_era5_europe_daily.py`
and `scripts/detect_heatwaves_europe.py` on a fresh checkout.


**Koppen class** (`koppen`): the climate type most common across the land the
event covered, whose first letter is the `zone` column.

| Code | Zone | Description | Where these events sat | Events |
|---|---|---|---|---|
| `Cfb` | temperate | Oceanic: mild, damp, no dry season | Ireland, Britain, northern France | 9 |
| `Dfb` | cold (continental) | Humid continental with warm summers | Central Europe, western Russia, southern Scandinavia | 8 |
| `Csa` | temperate | Mediterranean with hot, dry summers | Southern Italy, Greece, southern Spain, northern Algeria, Anatolia | 6 |
| `Dfc` | cold (continental) | Subarctic: short cool summers | Northern Norway and Sweden, Arctic Russia | 4 |
| `Csb` | temperate | Mediterranean with milder summers | Northern Portugal, the Spanish interior uplands | 2 |
| `Cfa` | temperate | Humid subtropical: hot summers, rain year round | Western Bulgaria | 1 |
| `Dsb` | cold (continental) | Continental with a dry summer | Eastern Turkish highlands | 1 |
| `BSk` | arid | Cold semi-arid steppe | Interior Algeria | 1 |

That is 18 temperate, 13 cold continental and 1 arid; the domain holds no
tropical or polar land, so two of the five zones never appear.

Two warnings. Five events are not European, 8 and 12 in Algeria and 22, 24 and
26 in Anatolia, and they hold the catalogue's two hottest readings of 49.2C and
48.0C. The top four each span more than 1,000 km from centroid to hottest cell,
so do not name them after one country.

**Centroid** - It is the area-weighted mean position over the cells in the event on its peak day, $\varphi_c=\sum A_{ij}\varphi_i/\sum A_{ij}$, and likewise for longitude.


## Setup

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.colors import LogNorm, Normalize

YEAR = 2026
# Repository root, wherever the notebook is opened from.
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "paper").is_dir())
RESULTS, PLOTS = ROOT / "paper" / "results", ROOT / "paper" / "figures"
# Downloads and thresholds live in the hwcat working folder.
WORKDIR = Path(os.environ.get("HWCAT_WORKDIR", ROOT / "hwcat-data"))
PLOTS.mkdir(parents=True, exist_ok=True)

# hwcat.plots calls mpl_apply() on import: the project rcParams come from
# there. Do not set rcParams here.
from hwcat import detect as D, plots as P  # noqa: E402

events = pd.read_csv(RESULTS / f"heatwave_events_europe_{YEAR}.csv",
                     parse_dates=["start", "peak", "end"])
sev = events["severity_C_km2_day"]
norm = LogNorm(vmin=sev.min(), vmax=sev.max())
cmap = plt.get_cmap("inferno_r")

print(f"{len(events)} events, {events.start.min():%d %b} to {events.end.max():%d %b}")
events.head()
```

## The catalogue

All 32 events with the columns that describe what kind of heat each one was,
and where. The two temperature columns disagree on purpose: `peak_tmax_C` is an
absolute reading and `max_anom_C` is degrees past the local threshold, so the
hottest event and the most extreme-for-its-place event are not the same row.
Figure 4 is this table as a picture.

## Figure 1: where the events were

The shading is the number of days each grid cell spent inside any detected
event, so all 32 contribute to it. `scripts/detect_heatwaves_europe.py` saves
the event table but discards `det.labels`, so the detection is rerun here
rather than read back. On this domain that costs about four seconds, which is
cheaper than maintaining a cache file.

The script's own version of this map labels only the 8 most severe events, and
its labels collide in southeast France. This one numbers all 32 and moves the
dates into a key.

```python
# The threshold the 2026 catalogue is built with (April-October, 15 days,
# full 1991-2020 baseline); `hwcat detect --year 2026` makes it if missing.
target = D.open_daily_tmax([WORKDIR / "era5" / f"europe_tmax_{YEAR}.nc"])
THRESH = WORKDIR / "thresholds" / "heatwave_threshold_europe_1991_2020_p90_m04-10.nc"
if not THRESH.exists():
    from hwcat.catalogue import run
    from hwcat.workdir import Workdir
    run(YEAR, Workdir(WORKDIR), plot=False)
thresh = xr.open_dataarray(THRESH).load()
land = D.land_mask_for(target["latitude"].values, target["longitude"].values)
det = D.detect(target, thresh, land_mask=land,
               min_days=3, min_area_km2=50_000.0)
hw_days = (det.labels > 0).sum(axis=0)

print(f"{det.n_events} blobs before the area filter; "
      f"busiest cell spent {hw_days.max()} days in an event")
```

```python
# top_n=0 draws the shaded map with no annotations. The paper's numbering
# and key are added below so all 32 events appear, not the default 8.
fig, ax = P.plot_europe_heatwaves(hw_days, det.lat, det.lon, events=events,
                                  top_n=0, year=YEAR)

# ponytail: greedy nudge, O(n^2) over 32 points. Swap for a real label solver
# only if the catalogue grows past a few hundred events.
placed = []
for _, e in events.iterrows():
    x, y = e.centroid_lon, e.centroid_lat
    while any(abs(x - px) < 1.7 and abs(y - py) < 1.3 for px, py in placed):
        y += 1.3
    placed.append((x, y))
    if y != e.centroid_lat:          # leader line back to the true centroid
        ax.plot([e.centroid_lon, x], [e.centroid_lat, y], color="black",
                linewidth=0.4, zorder=4)
    ax.scatter([e.centroid_lon], [e.centroid_lat], s=4, c="black", zorder=5)
    ax.annotate(str(int(e["rank"])), (x, y), fontsize=5.5, zorder=6,
                ha="center", va="center", color="black",
                bbox=dict(boxstyle="circle,pad=0.13", facecolor="white",
                          edgecolor="black", linewidth=0.5))

ax.set_title(f"European heat waves {YEAR}: days above the calendar-day "
             f"90th percentile\n(CTX90pct, 1991--2020 baseline; all "
             f"{len(events)} events numbered by severity rank)")

# Event key, parked in the empty Atlantic rather than outside the axes, which
# would otherwise leave a third of the canvas white.
key = [f"{int(e['rank']):2d}. {e.start:%d %b}--{e.end:%d %b} {e.koppen}"
       for _, e in events.iterrows()]
ax.text(-24.3, 50.2, "\n".join(key), fontsize=5.0, va="top", ha="left",
        linespacing=1.5, zorder=7)

fig.savefig(PLOTS / f"europe_{YEAR}_fig1_map.pdf", bbox_inches="tight")
```

## Figure 1b: the same map with country borders

Figure 1 carries no borders, which is honest about the data (the detector never
sees a country) but makes it hard to say where an event was. This draws the same
shading over national outlines so the events can be read off by country.

The outlines are Natural Earth 1:50m, clipped to the domain and rounded to two
decimals, which is 318 KB of plain GeoJSON in `data/`. Drawing them is a loop
over rings, so the notebook still needs no cartopy or geopandas.

```python
import json
import urllib.request

BORDERS = Path(P.__file__).parent / "data" / "borders_europe_ne50m.json"
if not BORDERS.exists():          # one-off fetch, ~3 MB down, 318 KB kept
    SRC = ("https://raw.githubusercontent.com/nvkelso/natural-earth-vector/"
           "master/geojson/ne_50m_admin_0_countries.geojson")
    W, S, E, N = -25.0, 35.0, 45.0, 72.0
    with urllib.request.urlopen(SRC, timeout=60) as r:
        world = json.load(r)
    rings = []
    for feat in world["features"]:
        geom = feat["geometry"]
        raw = (geom["coordinates"] if geom["type"] == "Polygon"
               else [r for poly in geom["coordinates"] for r in poly])
        for ring in raw:
            xs, ys = [p[0] for p in ring], [p[1] for p in ring]
            if max(xs) < W or min(xs) > E or max(ys) < S or min(ys) > N:
                continue          # ring lies wholly outside the domain
            rings.append([[round(x, 2), round(y, 2)] for x, y in ring])
    BORDERS.write_text(json.dumps(rings, separators=(",", ":")))
    print(f"fetched {len(rings)} rings -> {BORDERS}")

rings = json.loads(BORDERS.read_text())

fig, ax = P.plot_europe_heatwaves(hw_days, det.lat, det.lon, events=events,
                                  top_n=0, year=YEAR)
for ring in rings:
    ax.plot([p[0] for p in ring], [p[1] for p in ring],
            color="#333333", linewidth=0.35, zorder=3)

# Same numbering as Figure 1, so the two read together.
placed = []
for _, e in events.iterrows():
    x, y = e.centroid_lon, e.centroid_lat
    while any(abs(x - px) < 1.7 and abs(y - py) < 1.3 for px, py in placed):
        y += 1.3
    placed.append((x, y))
    if y != e.centroid_lat:
        ax.plot([e.centroid_lon, x], [e.centroid_lat, y], color="black",
                linewidth=0.4, zorder=4)
    ax.scatter([e.centroid_lon], [e.centroid_lat], s=4, c="black", zorder=5)
    ax.annotate(str(int(e["rank"])), (x, y), fontsize=5.5, zorder=6,
                ha="center", va="center", color="black",
                bbox=dict(boxstyle="circle,pad=0.13", facecolor="white",
                          edgecolor="black", linewidth=0.5))

ax.set_xlim(det.lon.min(), det.lon.max())
ax.set_ylim(det.lat.min(), det.lat.max())
ax.set_title(f"European heat waves {YEAR} over national borders\n"
             f"(days above the calendar-day 90th percentile; all "
             f"{len(events)} events numbered by severity rank)")

fig.savefig(
    PLOTS / f"europe_{YEAR}_fig1b_map_borders.pdf", bbox_inches="tight")
```

## Figure 2: when they ran

One bar per event from start to end, ordered by start date, shaded by severity.
The tick inside each bar is the peak day, the day with the highest mean
exceedance across the cells the event covered.

This is the figure that shows overlap: events that share a fortnight but not a
region are separate rows here and separate blobs on Figure 1.

```python
season = events.sort_values("start").reset_index(drop=True)

fig, ax = plt.subplots(figsize=(7.2, 6.4))
for y, (_, e) in enumerate(season.iterrows()):
    ax.barh(y, e.end - e.start + pd.Timedelta(days=1), left=e.start,
            height=0.62, color=cmap(norm(e.severity_C_km2_day)),
            edgecolor="black", linewidth=0.4)
    ax.plot(e.peak, y, marker="|", color="white", ms=5, mew=0.9, zorder=4)

ax.set_yticks(range(len(season)))
ax.set_yticklabels([f"{int(e['rank'])}. {e.koppen}"
                    for _, e in season.iterrows()], fontsize=6.5)
ax.invert_yaxis()
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
ax.grid(axis="x", alpha=0.3, linewidth=0.6)
ax.set(xlabel=f"{YEAR}", ylabel="severity rank and Koppen class",
       title=f"When the {len(events)} events ran\n"
       f"(bar spans start to end, tick marks the peak day)")
fig.colorbar(plt.cm.ScalarMappable(norm=norm, cmap=cmap), ax=ax,
             label=r"severity ($^\circ$C km$^2$ day)", pad=0.02, shrink=0.85)

fig.savefig(PLOTS / f"europe_{YEAR}_fig2_timeline.pdf", bbox_inches="tight")
```

## Figure 3: how severity is distributed?

Severity spans three orders of magnitude, so the bars need a log axis; on a
linear one the top event is the only visible bar. The blue curve is the
cumulative share, and it is the number to quote: it tells a reader how far down
the catalogue they need to go before the rest stops mattering.

**Severity** (`severity_C_km2_day`): the total excess heat an event delivered,
one number for how much of Europe was how far past normal for how long, so it
ranks events by overall burden and not by how hot it got anywhere.

$$S=\sum_{(t,i,j)\,\in\,E}\bigl[T_{\max}(t,i,j)-T_{90}(d_t,i,j)\bigr]\,A(i)$$

$E$ is the set of cell-days in the event, $T_{90}$ is that cell's 90th
percentile for that calendar day over 1991--2020, and $A(i)$ is the cell's area.

```python
cum = 100 * sev.cumsum() / sev.sum()

fig, ax = plt.subplots(figsize=(7.2, 4.4))
ax.bar(events["rank"], sev, color=cmap(norm(sev)), edgecolor="black",
       linewidth=0.4, width=0.75)
ax.set_yscale("log")
ax.set(xlabel="severity rank", ylabel=r"severity ($^\circ$C km$^2$ day)",
       title=f"Severity is concentrated: {len(events)} events, "
       f"three carry {cum.iloc[2]:.0f}\\% of the year")

ax2 = ax.twinx()
ax2.plot(events["rank"], cum, color="#1f6feb", marker="o", ms=2.6, lw=1.3)
ax2.set_ylabel(r"cumulative share of total severity (\%)", color="#1f6feb")
ax2.tick_params(axis="y", colors="#1f6feb")
ax2.set_ylim(0, 105)
ax2.grid(False)
for k in (3, 8):
    ax2.annotate(f"top {k}: {cum.iloc[k - 1]:.0f}\\%", (k, cum.iloc[k - 1]),
                 xytext=(10, -4), textcoords="offset points", fontsize=7.5,
                 color="#1f6feb", bbox=dict(facecolor="white", alpha=0.85,
                                            edgecolor="none", pad=1.2))

fig.savefig(PLOTS / f"europe_{YEAR}_fig3_severity.pdf", bbox_inches="tight")
```

## Figure 4: absolute heat against relative heat

`peak_tmax_C` is a temperature and `max_anom_C` is degrees above that cell's own
calendar-day 90th percentile. They rank the catalogue differently, and this
figure is how a reader decides which column their question needs. Colour is the
centroid latitude, so the north-south structure is visible directly.

```python
fig, ax = plt.subplots(figsize=(6.8, 5.2))
sc = ax.scatter(events.peak_tmax_C, events.max_anom_C,
                s=18 + 170 * events.peak_area_km2 / events.peak_area_km2.max(),
                c=events.centroid_lat, cmap="viridis", norm=Normalize(35, 70),
                edgecolor="black", linewidth=0.5, alpha=0.9)
fig.colorbar(sc, ax=ax, label=r"centroid latitude ($^\circ$N)", pad=0.02)

# Every point carries its rank. ponytail: greedy nudge, O(n^2) over 32
# points, same as Figure 1; a real label solver is not worth it at this size.
sizes = 18 + 170 * events.peak_area_km2 / events.peak_area_km2.max()
placed = []
for (_, e), s in zip(events.iterrows(), sizes):
    dx = 0.22 + 0.05 * np.sqrt(s)           # clear the marker, which varies
    x, y = e.peak_tmax_C + dx, e.max_anom_C
    while any(abs(x - px) < 1.6 and abs(y - py) < 0.40 for px, py in placed):
        y += 0.40
    placed.append((x, y))
    if abs(y - e.max_anom_C) > 1e-9:
        ax.plot([e.peak_tmax_C, x], [e.max_anom_C, y], color="0.45",
                linewidth=0.4, zorder=2)
    ax.annotate(int(e["rank"]), (x, y), fontsize=6.5, va="center",
                ha="left", zorder=6)

ax.grid(alpha=0.3, linewidth=0.6)
ax.set(xlabel=r"peak daily maximum temperature ($^\circ$C)",
       ylabel=r"largest exceedance above the local threshold ($^\circ$C)",
       title="Hot in absolute terms is not the same as extreme for the place\n"
             "(marker area scales with peak extent; numbers are severity ranks)")

fig.savefig(PLOTS / f"europe_{YEAR}_fig4_anomaly_vs_tmax.pdf",
            bbox_inches="tight")
```

## Interesting
### Figure 5: which climates the catalogue is made of

The glossary table as a picture, counted two ways. Counting events says the
catalogue is mostly oceanic; weighting by severity says it is mostly humid
continental. Both are true, and the gap between the panels is the point: the
common climate type here produced many modest events, while a less common one
produced the few that dominate the year.

```python
SHORT = {"Cfb": "oceanic", "Dfb": "humid continental",
         "Csa": "Mediterranean, hot", "Dfc": "subarctic",
         "Csb": "Mediterranean, mild", "Cfa": "humid subtropical",
         "Dsb": "continental, dry summer", "BSk": "cold semi-arid",
         "Dfa": "humid continental, hot summer"}
ZONE_COLOR = {"cold (continental)": "#4C72B0", "temperate": "#55A868",
              "arid": "#DD8452"}

by_class = (events.groupby(["koppen", "zone"])
            .agg(n=("rank", "size"), severity=("severity_C_km2_day", "sum"))
            .reset_index().sort_values("n"))
by_class["share"] = 100 * by_class.severity / sev.sum()
colors = [ZONE_COLOR[z] for z in by_class.zone]
y = np.arange(len(by_class))

fig, (a1, a2) = plt.subplots(1, 2, figsize=(8.4, 4.0), sharey=True)
a1.barh(y, by_class.n, color=colors, edgecolor="black", linewidth=0.4)
a1.set(xlabel="events", xticks=range(0, 11, 2))
for yi, n in zip(y, by_class.n):
    a1.text(n + 0.15, yi, str(n), va="center", fontsize=7.5)

a2.barh(y, by_class.share, color=colors, edgecolor="black", linewidth=0.4)
a2.set(xlabel=r"share of total severity (\%)")
for yi, s in zip(y, by_class.share):
    a2.text(s + 1.0, yi, f"{s:.1f}", va="center", fontsize=7.5)

a1.set_yticks(y)
a1.set_yticklabels([f"{k}  {SHORT[k]}" for k in by_class.koppen], fontsize=8)
for ax_ in (a1, a2):
    ax_.grid(axis="x", alpha=0.3, linewidth=0.6)
    ax_.set_axisbelow(True)
    ax_.margins(x=0.12)

handles = [plt.Rectangle((0, 0), 1, 1, facecolor=c, edgecolor="black",
                         linewidth=0.4) for c in ZONE_COLOR.values()]
fig.legend(handles, ZONE_COLOR.keys(), loc="lower center", ncol=3,
           frameon=False, fontsize=8, bbox_to_anchor=(0.5, -0.04))
fig.suptitle("Koppen classes in the 2026 catalogue, counted two ways", y=1.0)

fig.savefig(PLOTS / f"europe_{YEAR}_fig5_koppen.pdf", bbox_inches="tight")
```

## Figure 6: the footprints of the top 8

One panel per event, shading the number of days each cell spent inside that
event. Figure 1 is this sum taken over all 32 events at once; here it is taken
one event at a time, from the same `det.labels` array. The star is the cell
that recorded the event's `peak_tmax_C`, which for the largest events sits far
from the centroid.

Every panel uses the full domain of Figure 1, so a footprint's size and the
countries under it read directly against the other seven.


```python
TOP = 8
top = events.head(TOP)
lat, lon = det.lat, det.lon
extent = [lon.min(), lon.max(), lat.min(), lat.max()]
vmax = max(int((det.labels == eid).sum(axis=0).max()) for eid in top.event_id)

fig, axes = plt.subplots(4, 2, figsize=(7.4, 11.0), sharex=True, sharey=True)
fig.subplots_adjust(hspace=0.3, wspace=0.08)
for ax, (_, e) in zip(axes.flat, top.iterrows()):
    days = (det.labels == int(e.event_id)).sum(axis=0)
    im = ax.imshow(np.where(days > 0, days, np.nan), origin="upper",
                   cmap="inferno_r", vmin=1, vmax=vmax, extent=extent,
                   interpolation="nearest", aspect="auto", zorder=2)
    for ring in rings:
        ax.plot([p[0] for p in ring], [p[1] for p in ring],
                color="#333333", linewidth=0.3, zorder=3)
    ax.plot(e.peak_tmax_lon, e.peak_tmax_lat, marker="*", ms=8,
            mfc="#1f6feb", mec="white", mew=0.6, ls="none", zorder=5)

    ax.set_xlim(extent[0], extent[1])
    ax.set_ylim(extent[2], extent[3])
    ax.tick_params(labelsize=7)
    ax.set_title(f"{int(e['rank'])}. {e.start:%d %b} to {e.end:%d %b}, "
                 f"{e.koppen}\n{e.duration_days} d, "
                 f"{e.peak_area_km2 / 1e6:.2f} Mkm$^2$, "
                 f"{e.peak_tmax_C:.1f}$^\\circ$C", fontsize=7.5)

fig.colorbar(im, ax=axes, label="days inside the event", shrink=0.5,
             pad=0.02)
fig.suptitle(f"Footprints of the {TOP} most severe events of {YEAR}\n"
             "(star marks the hottest cell; all panels share the Figure 1 "
             "domain)", y=0.95)

fig.savefig(PLOTS / f"europe_{YEAR}_fig6_footprints.pdf", bbox_inches="tight")
```

## Summary for the write-up

Short bullets under the five headings of the brief paper. Numbers come from
the catalogue, `paper/results/catalogue_sensitivity_europe_2026.csv`, and like-for-like
runs of the same detector and threshold on 2003 and 2022
(`paper/results/heatwave_events_europe_2003.csv`, `..._2022.csv`).

**Method**

- ERA5 daily maximum 2 m temperature on the 0.25 degree grid, from the CDS
  daily-statistics catalogue, 1 May to 5 September 2026, 25W to 45E and 35N
  to 72N, land cells only.
- CTX90pct after Perkins and Alexander: a cell-day counts when Tmax exceeds
  that cell's calendar-day 90th percentile over 1991-2020, pooled over a
  15-day window (450 samples per threshold), for at least 3 consecutive days.
- Each day is filtered for spatial extent (50,000 km2 of connected land)
  before the space-time labelling, so scattered exceedances cannot bridge
  separate events. Events are connected components in (time, lat, lon).
- Severity is the exceedance summed over every cell-day, weighted by cell
  area: hot, large and long all count, so a one-day record does not outrank
  a continental month. Each event carries the modal Koppen class of the land
  it covered.

**Catalogue**

- 32 events from 1 May to 5 September; 18 temperate, 13 cold continental, 1
  arid. Five sit outside Europe (two in Algeria, three in Anatolia) and hold
  the two hottest readings, 49.2C and 48.0C.
- Severity is concentrated: the top three carry 58% of the year's total and
  the top eight 87%. Rank 1 (12 June to 3 July, central Europe, 22 days,
  2.98 Mkm2 at peak) is on its own worth 29%.
- The top four each span more than 1,000 km from centroid to hottest cell,
  so they are not one-country events. Ranks 1 and 2 are classed Dfb (humid
  continental), ranks 3 and 4 Cfb (oceanic): the year's heat sat over central
  Europe and France, not the Mediterranean.
- Sensitivity: ranks 1 to 5 survive every combination of minimum area (25k,
  50k, 100k km2) and percentile (90th, 95th) with the same dates within a
  day or two. At 25k the June and July waves (ranks 1 and 4) merge into one
  44-day event; at 100k rank 6 (Italy and the Balkans) breaks up and falls to
  rank 12. The 95th percentile shrinks events rather than removing them:
  rank 1 loses five days, ranks 2 and 3 swap.

**The three waves**

- The summer is three waves back to back over the same ground, together 58%
  of the year's severity: 12 June to 3 July over central Europe (rank 1),
  30 June to 24 July over France and the western Alps (rank 4), and 27 July
  to 19 August over southern France, Iberia and Italy (rank 2). Wave 2
  begins three days before wave 1 ends, and wave 3 begins three days after
  wave 2 ends.
- The 50,000 km2 floor is what separates waves 1 and 2; at 25,000 km2 they
  are one event. We can read them as one heat season with two pauses rather than
  three independent events.
- Bookends: a May precursor over France and the Alps (rank 3, 20 May to 2

## Figure 7: what survives a change of detector settings

The catalogue has two free parameters, the minimum daily patch area and the
percentile, and neither has a right answer. This tracks the six most severe
events of each season across all six combinations of them, for 2003, 2022 and
2026 from the same detector and the same 1991-2020 thresholds.

Read it as three rails. 2003 and 2022 stay flat, so their orderings are a
property of the season rather than of the settings. 2026 does not: rank 6 falls
to 12th, ranks 2 and 3 swap, and at 25,000 km2 the June and July waves are
detected as one 44-day event, which is the red ring. The mechanism is adjacency.
2026 delivered its severity as three waves back to back over the same ground, so
the area floor decides whether neighbouring waves are one event or two, while
2003's single wave and 2022's two separated centres have nothing to merge with.

The count above each column is the whole catalogue at that setting. Across the
three seasons it moves by a factor of 2.6 to 4.4, so the number of heat waves in
a season is largely a statement about the detector.

A bump chart rather than a grouped bar chart of counts, because the question is
whether the ORDER changes, and a merge then draws itself as two lines meeting.
The trace breaks between the two percentile groups: they are separate
experiments, not a sequence, and a connecting segment would imply a transition
that does not exist.

Needs `data/catalogue_sensitivity_europe_{2003,2022,2026}.csv`, from
`python scripts/catalogue_sensitivity.py --year <year>`.

```python
SENS_YEARS = (2003, 2022, 2026)
SETTINGS = [(25_000, 0.90), (50_000, 0.90), (100_000, 0.90),
            (25_000, 0.95), (50_000, 0.95), (100_000, 0.95)]
XPOS = np.array([0, 1, 2, 3.5, 4.5, 5.5])
N_REF = 6
SHAPE = {2003: "one August wave", 2022: "two separated centres",
         2026: "three waves over the same ground"}

sens = {y: pd.read_csv(RESULTS / f"catalogue_sensitivity_europe_{y}.csv")
        for y in SENS_YEARS}
ranks_cmap = plt.cm.viridis(np.linspace(0.0, 0.82, N_REF))

fig, axes = plt.subplots(3, 1, figsize=(7.4, 8.0), sharex=True)
fig.subplots_adjust(hspace=0.28)

for ax, year in zip(axes, SENS_YEARS):
    df = sens[year]
    ax.axvspan(XPOS[1] - 0.42, XPOS[1] + 0.42, color="0.88", zorder=0)

    # two reference events landing on one detected event = a merge
    merged = set()
    for (a, p), g in df.groupby(["min_area_km2", "percentile"]):
        for r in g.loc[g.duplicated("rank", keep=False), "rank"]:
            merged.add((a, p, int(r)))

    starts = {}
    for k in range(1, N_REF + 1):
        row = df[df.ref_rank == k].set_index(["min_area_km2", "percentile"])
        ys = np.array([float(row.loc[s, "rank"]) for s in SETTINGS])
        # The two percentile groups are separate experiments, not a sequence,
        # so the trace breaks between them rather than implying a transition.
        for sl in (slice(0, 3), slice(3, 6)):
            ax.plot(XPOS[sl], ys[sl], color=ranks_cmap[k - 1], lw=1.3,
                    marker="o", ms=4.2, mec="white", mew=0.6, zorder=3)
        starts.setdefault(ys[0], []).append(k)
        for s, x, y in zip(SETTINGS, XPOS, ys):
            if (s[0], s[1], int(y)) in merged:
                ax.plot(x, y, marker="o", ms=9, mfc="none", mec="#c1121f",
                        mew=1.1, zorder=5)

    # One label per starting rank: merged events share a point and would
    # otherwise print on top of each other.
    for y0, ks in starts.items():
        ax.annotate(",".join(str(k) for k in sorted(ks)), (XPOS[0], y0),
                    xytext=(-8, 0), textcoords="offset points", ha="right",
                    va="center", fontsize=8, color=ranks_cmap[min(ks) - 1],
                    zorder=4)

    trans = ax.get_xaxis_transform()
    for s, x in zip(SETTINGS, XPOS):
        n = int(df[(df.min_area_km2 == s[0])
                   & (df.percentile == s[1])].n_events.iloc[0])
        ax.text(x, 1.03, str(n), transform=trans, ha="center", va="bottom",
                fontsize=8, color="0.25")
    ax.text(-0.075, 1.03, "events:", transform=ax.transAxes, ha="right",
            va="bottom", fontsize=8, color="0.25")

    ax.set_ylim(13.2, 0.1)
    ax.set_yticks([1, 3, 6, 9, 12])
    ax.set_ylabel("rank")
    ax.grid(axis="y", alpha=0.3, lw=0.6)
    ax.set_title(f"{year}: {SHAPE[year]}", fontsize=9.5, loc="left", pad=16)

axes[-1].set_xticks(XPOS)
axes[-1].set_xticklabels(["25k", "50k", "100k"] * 2, fontsize=9)
axes[-1].set_xlim(-0.75, 6.1)
for x, lab in ((1.0, "90th percentile"), (4.5, "95th percentile")):
    axes[-1].text(x, -0.20, lab, transform=axes[-1].get_xaxis_transform(),
                  ha="center", va="top", fontsize=9)
axes[-1].text(2.75, -0.34, r"minimum daily patch area (km$^2$)",
              transform=axes[-1].get_xaxis_transform(), ha="center", va="top",
              fontsize=10)

fig.suptitle("Where the six most severe events of each season land under other "
             "detector settings\n"
             "(shaded column is the reference catalogue; red ring marks two "
             "reference events detected as one)", fontsize=10, y=0.975)

fig.savefig(PLOTS / "europe_sensitivity_ranks.pdf")
print("wrote", PLOTS / "europe_sensitivity_ranks.pdf")
```

## Figure 7b: the same result as a grid of rank changes

An alternative to Figure 7, kept alongside it so the choice can be made with a
draft in front of you. Same data, same claim, different encoding: colour carries
only how far an event moved, the printed number carries the direction, and an
unchanged cell is blank. Stable seasons therefore cost no ink, which is what
makes the three-year comparison land without tracing any lines.

Percentile is a column of the grid rather than a note in the caption. Figure 7
separates the two percentile blocks with a rule and explains it in the title,
which makes the reader take the key on trust; here each block sits under its own
heading.

**Sign convention.** The number is `ref_rank - new_rank`, not the other way
round, so a positive value means the event moved UP the ranking. That is the
only version that agrees with the figure: rank 1 sits at the top of the y axis,
so an event that falls moves down the page, and it would be contradictory to
print a positive number on it. 2026 rank 6 landing at 12th at 100,000 km2 is
therefore `-6`.

The blue bracket is the merge, and it needs the tie bar rather than two loose
outlines. At 25,000 km2 the June and July waves are one detected event, so
reference ranks 1 and 4 are both that event: rank 1 holds first place and shows
blank, and rank 4 has no separate event left to occupy, so it inherits first
place and reads `+3`. That is an artefact of the merge, not a promotion, which
is what the tie bar is there to say. Two unconnected boxes read as two unrelated
marks, which is how the first version was misread.

What it gives up is position. Figure 7 shows rank 6 of 2026 arriving at 12th as
a place on an axis; here that is the number `-6` and you have to read it.

The headline is narrower than "only 2026 reorders", which the cells contradict.
2003 and 2022 do reorder, among ranks 4 to 6 and only at the 95th percentile.
2026 reorders from rank 2 down and does it at both percentiles.

Reuses `SENS_YEARS`, `N_REF`, `SHAPE` and `sens` from the Figure 7 cell, so run
that one first.

```python
from matplotlib.patches import Rectangle

AREAS = (25_000, 50_000, 100_000)
PCTS = (0.90, 0.95)
VMAX = 6

fig, axes = plt.subplots(3, 2, figsize=(6.6, 6.2))
fig.subplots_adjust(hspace=0.75, wspace=0.13)
cmap = plt.get_cmap("YlOrRd").copy()
cmap.set_bad("white")              # unchanged really is blank

for row, year in enumerate(SENS_YEARS):
    df = sens[year]
    for col, pct in enumerate(PCTS):
        ax = axes[row, col]
        delta = np.zeros((N_REF, len(AREAS)))
        # Reference events that this setting detects as ONE event, grouped so
        # they can be tied together rather than outlined as unrelated cells.
        groups = []
        for j, area in enumerate(AREAS):
            g = df[(df.min_area_km2 == area) & (df.percentile == pct)]
            for _, r in g.iterrows():
                # Signed so it agrees with the page: rank 1 is at the
                # top, so positive means the event moved UP the
                # ranking and negative means it fell.
                delta[int(r.ref_rank) - 1, j] = int(r.ref_rank) - int(r["rank"])
            for _, tied in g[g.duplicated("rank", keep=False)].groupby("rank"):
                groups.append((j, sorted(int(k) - 1 for k in tied.ref_rank)))

        im = ax.imshow(np.ma.masked_where(delta == 0, np.abs(delta)),
                       cmap=cmap, vmin=0, vmax=VMAX, aspect="auto")
        for i in range(N_REF):
            for j in range(len(AREAS)):
                d = int(delta[i, j])
                if d:
                    ax.text(j, i, f"{d:+d}", ha="center", va="center",
                            fontsize=8, color="black" if abs(d) < 4 else "white")
        for j, rows in groups:
            for i in rows:
                ax.add_patch(Rectangle((j - .5, i - .5), 1, 1, fill=False,
                                       edgecolor="#1f6feb", lw=1.6, zorder=4))
            # The tie bar is what says "one event", which two loose outlines
            # do not: a reader asked what the blue meant.
            x = j - 0.42
            ax.plot([x, x], [min(rows) - .5, max(rows) + .5], color="#1f6feb",
                    lw=1.6, zorder=5, solid_capstyle="butt")
            ax.text(j - 0.34, np.mean(rows), "one\nevent", fontsize=6.5,
                    color="#1f6feb", ha="left", va="center", zorder=5)

        for j in range(1, len(AREAS)):
            ax.axvline(j - 0.5, color="white", lw=1.4)
        for i in range(1, N_REF):
            ax.axhline(i - 0.5, color="white", lw=1.4)

        counts = [int(df[(df.min_area_km2 == a)
                         & (df.percentile == pct)].n_events.iloc[0])
                  for a in AREAS]
        ax.set_xticks(range(len(AREAS)))
        ax.set_xticklabels(counts, fontsize=8.5)
        ax.tick_params(axis="both", length=0, pad=2)
        ax.set_yticks(range(N_REF))
        ax.set_yticklabels(range(1, N_REF + 1) if col == 0 else [],
                           fontsize=8.5)
        if col == 0:
            ax.set_ylabel("severity rank", fontsize=9)
        if row == len(SENS_YEARS) - 1:
            ax.set_xlabel("events in the whole season", fontsize=8.5,
                          labelpad=2)

        top = ax.secondary_xaxis("top")
        top.set_xticks(range(len(AREAS)))
        top.set_xticklabels([f"{a // 1000}k" for a in AREAS], fontsize=8.5)
        top.tick_params(length=0, pad=2)
        # Column heading on the top row only: the percentile is now a
        # structural axis of the grid, not something the caption has to say.
        # Placed in axes fractions above the year label, which sits at ~1.26.
        if row == 0:
            ax.text(0.5, 1.52, f"{int(pct * 100)}th percentile", fontsize=10,
                    transform=ax.transAxes, ha="center", va="bottom")

    # Year heading, left-aligned over the left panel so it reads as a row label
    axes[row, 0].text(-0.5, -1.55, f"{year}: {SHAPE[year]}", fontsize=9.5,
                      ha="left", va="bottom", fontweight="normal")

cb = fig.colorbar(im, ax=axes, fraction=0.03, pad=0.03)
cb.set_label("places moved", fontsize=9)   # magnitude; the
# printed sign carries direction, because the claim is that an
# event moved at all, not which way.
fig.suptitle("At the 95th percentile the August 2003 wave drops to second; "
             "2022 reorders only its 6th event.\nIn 2026 the reordering reaches "
             "rank 2 and happens at both percentiles.\n"
             "(blank = unchanged; a positive number means the event moved UP "
             "the ranking;\nblue ties reference events that this setting "
             "detects as a single event;\ncolumn heads are the minimum patch "
             "area)",
             fontsize=10, y=1.13)

fig.savefig(PLOTS / "europe_sensitivity_rank_changes.pdf",
            bbox_inches="tight")
print("wrote", PLOTS / "europe_sensitivity_rank_changes.pdf")
```
